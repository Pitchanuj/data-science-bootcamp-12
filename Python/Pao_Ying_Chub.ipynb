# **HW1 : Pao Ying Chub**
The `play_pyc()` function implements a classic game of Rock, Paper, Scissors (locally known in Thailand as Pao Ying Chub). It allows a user to play multiple rounds against the computer until they choose to exit

```python
## 1. Pao Ying Chub
import random

def report_score(user_score, comp_score, end = False):
    print("--------score----------")
    print(f"Computer: {comp_score} | You: {user_score}")
    print("-----------------------")
    if end:
        if comp_score > user_score:
            print("Game Result: Computer Wins")
        elif user_score > comp_score:
            print("Game Result: You win")
        else:
            print("Game Result: Tie")

def play_pyc():
    hands = ["rock", "paper", "scissor"]
    comp_score = 0
    user_score = 0
    count = 1
    print("Welcome to Pao Ying Chub Game")
    print("------------------------------")

    while True:
        print(f"Round: {count}")
        print("Please choose the following option to play : \n"
        "[1]Rock [2]Paper [3]Scissor [0]Exit")
        try:
            user_hand = int(input("Enter your choice: "))
            if user_hand not in [0,1,2,3]:
                print("❌Please Enter 1, 2, 3 or 0")
                continue
        except ValueError:
            print("❌Please Enter 1, 2, 3 or 0")
            continue

        if user_hand == 0:
            report_score(user_score, comp_score, end = True)
            break

        comp_hand = random.randint(1,3)

        print(f"You: {hands[user_hand-1]}")
        print(f"Computer: {hands[comp_hand-1]}")

        if user_hand == comp_hand:
            print("Tie")
        elif (user_hand == 1 and comp_hand == 3) or \
            (user_hand == 2 and comp_hand == 1) or \
            (user_hand == 3 and comp_hand == 2):
            print("You win this round!")
            user_score += 1
        else:
            print("Computer wins this round")
            comp_score += 1

        report_score(user_score, comp_score)
        count += 1
```

```python
play_pyc()
```
