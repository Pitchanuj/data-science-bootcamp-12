# HW2: Pizza Restaurant
This `order_pizza()` function acts as a Point of Sale (POS) system for a pizza restaurant. It manages a dynamic ordering process, allowing users to select multiple items from the menus, calculates costs, and provides a final receipt

```python
def order_pizza():
    pizza_menu = {
            "hawaiian":{"s":120, "m":180, "l":240},
            "bbq meatlover":{"s":120, "m":180, "l":240},
            "margarita":{"s":100, "m":160, "l":220},
            "pepperoni":{"s":140, "m":200, "l":260},
            "seafood":{"s":180, "m":240, "l":280}
        }
    appetizer_menu = {
            "chicken wings":80,
            "french fries":50,
            "ceasar salad":80
    }
    drink_menu = {
            "coke" : 30,
            "lemonade":45,
            "mineral water":15
    }

    order_name = []
    order_price = []

    print("Welcome to our pizza restaurant")

    while True:
        option = input("Do you want to add order?(Y/N) ").lower().strip()
        if option not in ['n', 'y']:
            continue
        if option.lower() == 'n':
            print("-------------------")
            print("Your Order:")
            for i in order_name:
                print(i)
            print(f"Total: {sum(order_price)} THB")
            print("-------------------")
            break

        if option.lower() == 'y':
            print("[1] Pizza")
            print("[2] Appetizer")
            print("[3] Drink")
            while True:
                user_choice = input("What would you like to order? ").lower().strip()

                if user_choice in ["1", "pizza"]:
                    pizza_keys = list(pizza_menu.keys())
                    print("Toppings Option:")
                    for i in range(len(pizza_keys)):
                        print(f"[{i}] {pizza_keys[i]}")

                    while True:
                        try:
                            topping = input("What pizza topping do you want")
                            if topping in pizza_keys:
                                break
                            elif int(topping) in range(len(pizza_keys)):
                                topping = pizza_keys[int(topping)]
                                break
                            else:
                                print("Invalid value. Please choose a topping from the list.")
                        except ValueError:
                            print("Invalid value. Please choose a topping from the list.")

                    print("Size Option:")
                    for size, price in pizza_menu[topping].items():
                        print(f"Size {size} : {price} THB")
                    while True:
                        size = input("What size of pizza do you want (S, M, L)").lower().strip()
                        if size in ["s", "m", "l"]:
                            break
                        print("Invalid input. Please select s, m or l")

                    order_name.append(f"{topping.capitalize()} Pizza ({size.upper()}) - {pizza_menu[topping][size]} THB")
                    order_price.append(pizza_menu[topping][size])
                    print(f"{topping.capitalize()} Pizza ({size.upper()}) added")
                    break


                if user_choice in ["2", "appetizer"]:
                    for index, (name, price) in enumerate(appetizer_menu.items()):
                        print(f"[{index}] {name} : {price} THB")
                    while True:
                        try:
                            appetizer = input("What would you like to order?").lower().strip()
                            if appetizer in list(appetizer_menu.keys()):
                                break
                            elif int(appetizer) in range(len(list(appetizer_menu.keys()))):
                                appetizer = list(appetizer_menu.keys())[int(appetizer)]
                                break
                            print("Invalid value. Please choose appetizer from the list.")
                        except ValueError:
                            print("Invalid value. Please choose appetizer from the list.")
                    order_name.append(f"{appetizer} - {appetizer_menu[appetizer]}")
                    order_price.append(appetizer_menu[appetizer])
                    print(f"{appetizer.capitalize()} added")
                    break

                if user_choice in ["3", "drink"]:
                    for index, (name, price) in enumerate(drink_menu.items()):
                        print(f"[{index}] {name} : {price} THB")
                    while True:
                        try:
                            drink = input("What would you like to order?").lower().strip()
                            if drink in list(drink_menu.keys()):
                                break
                            elif int(drink) in range(len(list(drink_menu.keys()))):
                                drink = list(drink_menu.keys())[int(drink)]
                                break
                            print("Invalid value. Please choose drink from the list.")
                        except ValueError:
                            print("Invalid value. Please choose drink from the list.")
                    order_name.append(f"{drink} - {drink_menu[drink]}")
                    order_price.append(drink_menu[drink])
                    print(f"{drink.capitalize()} added")
                    break

                print("Invalid input. Please enter 1, 2, or 3 (or the name).")
```

```python
order_pizza()
```
