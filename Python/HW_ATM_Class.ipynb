# HW3: Create ATM Class
This class simulates an ATM where multiple accounts can interact in one system. It provides essential functions like account creation, deposits, withdrawals, and inter-user transfers. Every action is recorded in a history log

- 1. check_balance()
- 2. deposit()
- 3. withdraw()
- 4. transfer()
- 5. show_history()
- 6. close account()


```python
from datetime import datetime
from zoneinfo import ZoneInfo
```

```python
class ATM:
    accounts = {}

    @classmethod
    def list_accounts(cls):
        return list(cls.accounts.keys())

    @classmethod
    def get_time_th(cls):
        return datetime.now(ZoneInfo("Asia/Bangkok")).strftime("%Y-%m-%d %H:%M")

    def __init__(self, username, bank, amount):
        if username in ATM.accounts:
            raise ValueError("This username already exists")
        self.username = username
        self.bank = bank
        self.amount = amount
        self.history = []
        self.history.append(f"{ATM.get_time_th()} Account opened. Balance : {self.amount} THB")
        ATM.accounts[username] = self

    def __str__(self):
        return f"{self.username} uses {self.bank} and has {self.amount} THB in an account."

    # check balance
    def check_balance(self):
        print(f"Balance: {self.amount} THB")

    # deposit money into the account
    def deposit(self, saving):
        self.amount += saving
        print(f"Deposit {saving} THB succesfully")
        print(f"Your currect balance is {self.amount} THB")

        self.history.append(f"{ATM.get_time_th()} Deposit {saving} THB")

    # withdraw money from the account
    def withdraw(self, value):
        if value <= 0 :
            print("You entered invalid value")
        elif self.amount >= value:
            self.amount -= value
            print(f"Withdraw {value} THB successfully")
            print(f"Your currect balance is {self.amount} THB")

            self.history.append(f"{ATM.get_time_th()} Withdraw {value} THB")
        else:
            print(f"Your balance is insufficient")

    # trasfer money to target account (if existed)
    def transfer(self, target, value):
        if value <= 0:
            print("You entered invalid value")
        elif self.amount >= value:
            if target in ATM.accounts:
                self.amount -= value
                ATM.accounts[target].amount += value
                print(f"Transfered {value} to {target} successfully")
                print(f"Your currect balance is {self.amount} THB")

                self.history.append(f"{ATM.get_time_th()} Transfer {value} THB to {target}")
                ATM.accounts[target].history.append(f"{ATM.get_time_th()} Received {value} THB from {self.username}")
            else:
                print("Target account does not exist")
        else:
            print(f"Your balance is insufficient")

    # show transaction history
    def show_history(self):
        for i in self.history:
            print(i)

    # close currect account
    def close_account(self):
        del ATM.accounts[self.username]
        print(f"Withdraw {self.amount} THB")
        print(f"Account {self.username} closed.")
```

```python
jane = ATM("jane", "SCB", 500)
marry = ATM("marry", "KBank", 1000)
```

```python
# deposit money
marry.deposit(300)
```

```python
# withdraw money
marry.withdraw(200)
```

```python
marry.transfer("jane", 100)
```

```python
jane.check_balance()
```

```python
jane.show_history()
```

```python
marry.show_history()
```

```python
jane.close_account()
```
