# **HW4: Get Data From API**

We will get Disney character data from https://disneyapi.dev/ and convert into DataFrame

```python
# import library
import requests
import numpy as np
import pandas as pd
```

```python
url = "https://api.disneyapi.dev/character"

response = requests.get(url)
response
```

```python
response_json = response.json()
page = response_json['info']["totalPages"]
page
```

```python
characters = []
for i in range(page):
    url = f"https://api.disneyapi.dev/character?page={i+1}"
    response = requests.get(url).json()
    for character in response["data"]:
        data = [character["_id"],
            character["name"],
            ", ".join(character["films"]) if len(character["films"]) > 0 else np.nan,
            ", ".join(character["shortFilms"]) if len(character["shortFilms"]) > 0 else np.nan,
            ", ".join(character["tvShows"]) if len(character["tvShows"]) > 0 else np.nan,
            ", ".join(character["videoGames"]) if len(character["videoGames"]) > 0 else np.nan,
            len(character["films"]) + len(character["shortFilms"]) + len(character["tvShows"]) + len(character["videoGames"])]
        characters.append(data)
```

```python
characters_df = pd.DataFrame(characters, columns=["id", "name", "films", "shortFilms", "tvShows", "videoGames", "total_appearance"])
characters_df.head(10)
```

```python
characters_df.info()
```

```python
# write to csv
characters_df.to_csv("disney_characters.csv", index = False)
```

List the characters with the highest number of appearances on screen media

```python
top_media_character = characters_df[["name", "total_appearance"]].sort_values("total_appearance", ascending = False).head(10)
top_media_character = top_media_character.reset_index(drop = True)
top_media_character
```
